```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
k=np.arange(1, 11)

S=np.zeros(len(k))

def afunc(k):
    return 1/k**2

S[0] = afunc(1)

for i in range(1, len(k)):
    S[i] = S[i-1]+afunc(k[i])

plt.plot(k, S, 'o-', color='blue', label=r"$a=\frac{1}{k^2}$")
plt.xlim(1,10)
plt.xlabel('Number of terms')
plt.ylabel('Partial sum')
plt.legend()
plt.grid('on')
plt.show()
```

```python
# define the range of k values
k = np.arange(1, 11)

# uses cumsum to compute partial sums instead of a loop 
a = np.cumsum(1 / k ** 2)
b = np.cumsum((-1) ** (k + 1) / k)
c = np.cumsum(1 / np.sqrt(k))
d = np.cumsum(1 / (1 + np.log(k)))  # np.log is the natural logarithm (base e)
```

```python
plt.plot(k, a, "o-", color="blue", label=r"$a=\frac{1}{k^2}$")
plt.plot(k, b, "o-", color="orange", label=r"$b=\frac{(-1)^{k+1}}{k}$")
plt.plot(k, c, "o-", color="green", label=r"$c=\frac{1}{\sqrt{k}}$")
plt.plot(k, d, "o-", color="red", label=r"$d=\frac{1}{1+\ln(k)}$")
plt.xlim(1,10)
plt.xlabel('Number of terms (k)')
plt.ylabel('Partial sum')
plt.grid(True)
plt.legend()
plt.show()
```
