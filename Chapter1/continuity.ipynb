```python
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interactive
```

```python
def f(x):
    return 2/x

def finverse(x):
    return 2/x

x = np.linspace(0.5, 2)
x0 = 1
y = f(x)
y0 = f(x0)

def plot(eps):
    y0p = y0+eps
    y0m = y0-eps
    x0p = finverse(y0p)
    x0m = finverse(y0m)
    
    vertical = [x0, x0p, x0m]
    horizontal = [y0, y0p, y0m]
    
    plt.plot(x, y, 'r')   
    for Y in horizontal:
        plt.axhline(y = Y, color = 'k', 
                    linestyle = ':')    
    for X in vertical:
        plt.axvline(x = X, color = 'c', 
                    linestyle = ':')          
    plt.show()       
    
    delta= min(abs(x0-x0p), abs(x0-x0m))
    
    print(f'Given \u03B5 = {eps:.2}')
    print(f'Found \u03B4 = {delta:.4}')
    
interactive(plot, eps=(0.01, 0.4, 0.01))
```
